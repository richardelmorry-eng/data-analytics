## Lab 3. Updating an HR document

```python
# HR data: (employee#, department, name, salary)
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
hr_list
```

```python
# Change Mark's last name to Blick-Hawley (replace the tuple at index 1)
hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
print(hr_list)
```

```python
# Change Jim's department to CS and salary to 60000 (tuple at index 3)
hr_list[3] = ('0234', 'CS', 'Jim Smith', '60000')
print(hr_list)
```

```python
# Print each employee, casting salary to int and formatting with a comma
for emp in hr_list:
    salary = int(emp[3])
    print(f"{emp[0]} | {emp[1]} | {emp[2]} | {salary:,}")
```
